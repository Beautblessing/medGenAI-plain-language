```python
# ----- Semantic Preservation -----
from bert_score import score

def calculate_bertscore(original, simplified):
    """
    Calcula el BERTScore
    """
    P, R, F1 = score([simplified], [original], lang="es", rescale_with_baseline=True)
    return P.mean().item(), R.mean().item(), F1.mean().item()
```

```python
# ----- Readability for English -----
import textstat

def calculate_fkgl_fre(text):
    """
    Calcula FKGL (nivel de grado escolar) y FRE (Facilidad de lectura).
    """
    fkgl = textstat.flesch_kincaid_grade(text)
    fre = textstat.flesch_reading_ease(text)
    return fkgl, fre
```

```python
# ----- Readability for Spanish -----
import textstat

def calculate_fh(text):
    """
    Calcula Fernandez Huerta.
    """
    fh = textstat.flesch_kincaid_grade(text)
    return fh
```

```python
def evaluate_simplification(original, simplified, references):
    print("Evaluando simplificación de texto...\n")

    P, R, F1 = calculate_bertscore(original, simplified)
    print(f"BERTScore - Precisión: {P:.4f}, Recall: {R:.4f}, F1: {F1:.4f}")

    fkgl, fre = calculate_fkgl_fre(simplified)
    print(f"FKGL (Flesch-Kincaid Grade Level): {fkgl:.2f}")
    print(f"FRE (Flesch Reading Ease): {fre:.2f}")

    fh = calculate_fh(simplified)
    print(f"FR (Fernandez Huerta): {fh:.2f}")

if __name__ == "__main__":
    original_text = "Esta es una oración más fácil de entender Esta es una oración más fácil de entender.."
    simplified_text = "Esta es una oración más fácil de entender."
    references = ["Esta es una oración más fácil de entender."]


    evaluate_simplification(original_text, simplified_text, references)
```
